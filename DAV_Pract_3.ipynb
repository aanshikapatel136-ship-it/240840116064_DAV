**Step 1: Display Dataset**

```python
# Import Library.
import pandas as pd
# Create Dataset.
df = {
    "EMP_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Neha", "Vivek", "Pooja"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "HR"],
    "City": ["Surat", "Navasari", "Surat", "Valsad", "Surat", "Navasari", "Valsad", "Surat"],
    "Salary": [45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],
    "Experience": [2, 1, 5, 4, 2, 6, 8, 3]
}

df=pd.DataFrame(df)
df
```

**Step 2: Filter Data**

```python
# 2.1: Display employees from IT department.
df[df['Department'] == 'IT']
```

```python
# 2.2 Display employees with salary > 50000.
df[df['Salary'] > 50000]
```

```python
# 2.3 Display employees from Surat.
df[df['City'] == 'Surat']
```

```python
# 2.4 Display employees from IT department having salary > 5000
df[(df['Department'] == 'IT') & (df['Salary'] > 50000)]
```

**Step 3: Sort Data**

```python
# 3.1: Salary ascending
#sort employees by salary from lowest to highest
df.sort_values('Salary', ascending=True)
```

```python
# 3.2: Salary descending
#sort employees by salary from highest to lowest
df.sort_values('Salary', ascending=False)
```

```python
# 3.3: Department-wise and salary descending
#sort employee department wise and within each department sort salary in decending order
df.sort_values(by=['Department', 'Salary'],ascending=[True, False])
```

**Step 4: Perform Aggrigation**

```python
# 4.1 Calculate the total salary of all employees.
df['Salary'].sum()
```

```python
#4.2 Calculate thye avg salary of all employees.
df['Salary'].mean()
```

```python
# 4.3 Display max salary.
df['Salary'].max()
```

```python
# 4.4 Display lowest salary.
df['Salary'].min()
```

```python
# 4.5 Count the total no of employees.
df['EMP_ID'].count()
```

**Step 5: Perform GroupBy Operation**

```python
# 5.1: Average salary of each department
# grp employees based on department and calculate the avg salary of each department.
df.groupby('Department')['Salary'].mean().reset_index()
```

```python
# 5.2: Total salary of each department.
# Calculate the total salary paid by each dept.
df.groupby('Department')['Salary'].sum().reset_index()
```

```python
# 5.3: Highest salary in each department.
# Display the higest salary in each department.
df.groupby('Department')['Salary'].max().reset_index()
```

```python
# 5.4: Number of employees in each department.
# Count the no of employees in each dept.
df.groupby('Department')['EMP_ID'].count().reset_index()
```

```python
# 5.5: Average salary in each city.
# Calculate the avg salary of employee in each city.
df.groupby('City')['Salary'].mean().reset_index()
```

**Step 6: Multiple Aggrigation Function**

```python
# 6.1 Apply multiple statistical function to the salary column each dept.
df.groupby('Department')['Salary'].agg(['sum', 'mean', 'max', 'min', 'count']).reset_index()
```

**Step 7: Save the result**

```python
# Save CSV
df.to_csv("Employee_Data.csv", index=False)

print("CSV file saved Successfully")
```
