**Step 1: Import Required Libraries**

```python
# Import Pandas.
import pandas as pd

# Import Plotly Express.
import plotly.express as px

# Import Ploty Graph Objects.
import plotly.graph_objects as go
```

**Step 2: Create Dataset**

```python
data = {
    "Month": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December" ],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"],
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West", "North", "South", "East", "West"],
    "Sales": [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 2300010, 250000],
    "Units_Sold": [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40],
    "Profit": [18000, 27000, 30000, 22000, 32000, 35000, 28000, 34000, 38000, 33000, 46000, 50000]
}
df = pd.DataFrame(data)
df
```

**Step 3: Inspect the Dataset**

```python
# Display First Five Columns.
df.head()
```

```python
# Display No. of Rows and Columns.
df.shape
```

```python
# Display Information od dataset.
df.info()
```

**Step 4: Create an Interactive Line chart using Plotly Explore**

```python
# How does the sales revenue change throughout the year?
fig = px.line(df, x="Month", y="Sales", title="Sales Revenue Over Time")
fig.show()
```

**Step 5: Improve the Hover Information**

```python
# Create a Line Chart using a Sales over Month.
fig = px.line(df, x='Month', y='Sales', title='Sales Over Month', markers=True)
fig.show()
```

```python
# Add additional columns to the hover information. Student can see Product, Region, Units Sold and Profit when they move the mouse over a data point.
import plotly.express as px
fig = px.line(df,x="Month",y="Sales",markers=True,title="Monthly Sales Trend with Extra Hover Details",hover_data=["Profit", "Units_Sold", "Region"])
fig.show()
```

**Step 6: Create an Interective Bar Chart**

```python
# Group the datset accordimg to Product.
product_sales = df.groupby('Product')['Sales'].sum().reset_index()
```

```python
# Calculate the total Sales generated by each product.
df.groupby('Product')['Sales'].sum().reset_index()
```

```python
# Create an interective bar chart.
fig = px.bar(
    product_sales,
    x='Product',
    y='Sales',
    title='Total Sales by Product',
    labels={'Sales': 'Total Sales ($)', 'Product': 'Product Name'},
    text_auto='.2s'
)
fig.show()
```

**Step 7: Interective Scatter Plot**

```python
# TO analyze the relationship between Units Sold and Sales Using an Interective Scatter Plot.
fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    title="Units Sold vs Sales",
    labels={"Units_Sold": "Units Sold", "Sales": "Sales ($)"}
)
fig.show()
```

**Step 8: Interective Scatter Plot with Color**

```python
# Create an interactive scatter plot with color
fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    color="Product",
    title="Units Sold vs Sales by Product",
    labels={"Units_Sold": "Units Sold", "Sales": "Sales ($)"}
)
fig.show()
```
