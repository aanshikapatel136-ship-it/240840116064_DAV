Step 1: Import Pandas Library

```python
#import pandas library
import pandas as pd
```

Step 2: Create dataFram

```python
#create Dataset.
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krisha", "Harry"],
    "Age": [20, 21, 19, 20, 22, 24, 28 ],
    "City": ["Surat", "Navsari", "Valsad", "Bharuch", "Surat", "Gandhinagar", "Ahemdabad"],
    "Marks": [85, 92, 78, 88, 81, 84, 90]
}
```

Step 3: Create and Display Dataset

```python
#create Dataset.
df = pd.DataFrame(data)
#Display Dataset.
df
```

Step 4: Save DataFrame as CSV File

```python
#Convert Dataframe to CSV File.
df.to_csv("students.csv", index=False)
#Display Dataset.
print("CSV file created successfully.")
```

Step 5: Save DataFrame as Excel File

```python
#Convert Dataframe to Excel File.
df.to_excel("students.xlsx", index=False)

#Display Dataset.
print("Excel file created successfully.")
```

Step 6: Save DataFrame as JSON File

```python
#Convert Dataframe to JSON File.
df.to_json("students.json",orient="records", index=4)

#Display Dataset.
print("JSON file created successfully.")
```
