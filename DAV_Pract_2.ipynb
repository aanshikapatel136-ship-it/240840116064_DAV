Step 1: Import Pandas Library

```python
#import Pandas

import pandas as pd
```

Step 2: Create dataFram

```python
#create Dataset.
data = {
    "ID": [101, 102, 103, 104, 104, 105, 106, 107],
    "Name": ["Rahul", "Priya", "Amit", None, "Riya", "Ajay", "Naresh", "Jenil"],
    "Age": [20, 21, None, 20, 20, 22, 23, 28 ],
    "City": ["Surat", "Navsari", "Surat", "Bharuch","Bharuch", "Navsari", "Surat", "Ahemdabad"],
    "Marks": [85, None, 78, 88, 88, 81, 90, 93]
}

df = pd.DataFrame(data)
df
```

Step 4: Check Missing Values

```python
#Check for the missing values in every columns.

df.isnull()
```

```python
#Print null values for each columns.

df.isnull().sum()
```

```python
#print total number of null values in dataset

df.isnull().sum().sum()
```

Step 5: Fill the missing values

```python
# Replace the missing values in the Marks column.

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

#Display the updated DataFrame.
df
```

```python
#Replace Missing value in the Age column.

df["Age"] = df["Age"].fillna(df["Age"].mean())
#Display the updated DataFrame.
df
```

```python
#Replace the missing Names with the text "unknown".
df["Name"] = df["Name"].fillna("Unknown")
#Display the updated DataFrame.
df
```

Step 6:Detect the duplicate value.


```python
#Detect the duplicate value.
#Display the updated DataFrame.
df.duplicated()
```

```python
#Count the total number of duplicate rows.
df.duplicated().sum()
#Display the updated DataFrame.
df
```

Step 7: Remove the duplicate records

```python
# Delete duplicate data.
df.drop_duplicates(inplace=True)
#Display the updated DataFrame.
df
```

Step 8: rename the column

```python
#rename the column "Marks" to "Score"
df.rename(columns={"Marks": "Score"}, inplace=True)
#Display the updated DataFrame.
df
```

Step 9: Chnage Data Type

```python
#Convert the age column from float to integer.
df["Age"] = df["Age"].astype(int)
#Display the updated DataFrame.
df.dtypes
```

Step 10: Create a new column

```python
#Create a new column names "Final Score".
df ["Final Score"] = df["Score"] + 5
#Display the updated DataFrame.
df
```

Step 11: Convert text to upper case

```python
#Student Name column to uppercase
df["Name"] = df["Name"].str.upper()
#Display the updated DataFrame.
df
```

Step 12: Convert text into Lowercase

```python
#Convert City into lowercase.
df["City"] = df["City"].str.lower()
#Display the updated DataFrame.
df
```

Step 13: Remove Extra Space

```python
#Remove Extra space form city column.
df["City"] = df["City"].str.strip()
#Display the updated DataFrame.
df
```

Step 14: Sort the Dataset

```python
#Sort the dataset based on final score.
df.sort_values(by=["Final Score"])
#Display the updated DataFrame.
df
```

Step 15:Save the CleanData

```python
# Save the cleaned DataFrame
df.to_csv('Cleaned_Students.csv', index=False)

# Display the updated DataFrame
df
```
